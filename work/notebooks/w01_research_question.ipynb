## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Question:** Given a page's current signals (position, CTR, impressions,
word count, content age), how likely is it to be declining, and does that
likelihood justify prioritizing it for a content refresh?

**Unit of analysis:** one page, scored at a point in time.

**Decision it improves:** which pages an SEO or content team should review
first out of a large backlog.

**Action taken:** a content strategist pulls the top-N ranked pages and
schedules them for a refresh sprint.

**Cost of a wrong call:**
- False positive (flagged as declining, wasn't): wastes limited refresh
  hours on a page that didn't need it.
- False negative (missed a real decliner): the page keeps losing traffic
  silently until the next review cycle.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Three numbers from the starter dataset that support this lane:

```python
import pandas as pd
import json
import os, subprocess, sys

# Make sure the pipeline has been run so outputs exist
if not os.path.exists("outputs/model_results.json"):
    subprocess.run([sys.executable, "scripts/run_all.py"], check=True)

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: base rate of declining pages
decline_rate = (df["trend_direction"] == "down").mean()
print(f"Share of pages currently declining: {decline_rate:.1%}")

# Number 2: model lift over the hand-written baseline
res = json.load(open("outputs/model_results.json"))
base = res["baseline"]["baseline_precision_at_50"]
rf = res["models"]["random_forest"]["precision_at_50"]
print(f"Baseline rule Precision@50: {base:.2f}")
print(f"Model Precision@50: {rf:.2f}")
print(f"Lift: {rf / base:.1f}x")

# Number 3: decline rate varies by position tier
tier_decline = df.groupby("position_tier")["trend_direction"].apply(
    lambda x: (x == "down").mean()
)
print("\nDecline rate by position tier:")
print(tier_decline.round(2).to_string())
```
